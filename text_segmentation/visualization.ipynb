```python
import json

windiffs = {}
pks = {}
for model in ('bertseg', 'treeseg'):
    with open(f'metrics_{model}.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
    windiffs[model] = [item['windiff'] for item in data]
    pks[model] = [item['pk'] for item in data]
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.boxplot([windiffs['bertseg'], windiffs['treeseg'], pks['bertseg'], pks['treeseg']], tick_labels=['windiff bertseg', 'windiff treeseg', 'pk bertseg',  'pk treeseg'])
plt.ylim(0, 1)
plt.title('Distribution de windiff et pk')
plt.ylabel('Valeur')
plt.show()
```

```python
import numpy as np

metrics = {
    'Windiff BertSeg': windiffs['bertseg'],
    'pk BertSeg': pks['bertseg'],
    'Windiff TreeSeg': windiffs['treeseg'],
    'pk TreeSeg': pks['treeseg']
}

for name, values in metrics.items():
    print(f"\n{name}:")
    print(f"Mean: {np.mean(values):.3f}")
    print(f"Median: {np.median(values):.3f}")
    print(f"Min: {np.min(values):.3f}")
    print(f"Max: {np.max(values):.3f}")
```
